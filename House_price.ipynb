## Dropping Address and State

### Why Address is dropped
- **Address** is a free-text field, and every one of the 4,548 rows has a unique value.
- A column where every value is different works like an ID. A model cannot learn a pattern from it, so it adds no predictive value.
- The street, city and zip code are randomly generated, so they carry no real information about price.
- Keeping it would only add noise and make the dataset harder to encode.

### Why State is dropped
- **State** was extracted from **Address** using a regex, to test whether location affects price.
- The average price per state ranges from about 1.14M (HI) to about 1.35M (TX), a max/min ratio of 1.175.
- With only about 70 houses per state across ~62 categories, a gap of this size is expected from random variation alone.

- **Qn.** Do more rooms or bedrooms raise the price?

```python
# Outlier Detection
outlier_summary = []

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    n_out = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        'Column': col,
        'Lower Bound': round(lower, 2),
        'Upper Bound': round(upper, 2),
        'Min': round(df[col].min(), 2),
        'Max': round(df[col].max(), 2),
        'Outliers': n_out,
        'Outlier %': round(n_out / len(df) * 100, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df
```

```python
# 2. Boxplots for every numeric column
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    sns.boxplot(y=df[col], ax=axes[i], color='steelblue')
    axes[i].set_title(col, fontweight='bold')

fig.suptitle('Outlier Detection (Boxplots)', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
