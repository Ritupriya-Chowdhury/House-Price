```python
# Convert "Number of Rooms" and "Number of Bedrooms" float to int
df[['Number of Rooms', 'Number of Bedrooms']]=df[['Number of Rooms', 'Number of Bedrooms']].astype(int)
```

```python
# Similarly " House Age" & "Area Population"
df['House Age'] = df['House Age'].round().astype(int)
df['House Age'] = df['House Age'].round(1)
```

```python
(df['Number of Bedrooms'] > df['Number of Rooms']).sum()
```

```python
df=df[df['Number of Bedrooms'] <= df['Number of Rooms']].reset_index(drop=True)
```
